```python

class Employee:

    def __init__(self, employee_id, name, age):
        self.employee_id = employee_id
        self.name = name
        self.age = age

    def display_info(self):
        print("Employee ID:", self.employee_id)
        print("Name:", self.name)
        print("Age:", self.age)

    def calculate_salary(self):
        pass



class FullTimeEmployee(Employee):

    def __init__(self, employee_id, name, age, monthly_salary):
        super().__init__(employee_id, name, age)
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        return self.monthly_salary



class PartTimeEmployee(Employee):

    def __init__(self, employee_id, name, age, hourly_rate, hours_worked):
        super().__init__(employee_id, name, age)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        return self.hourly_rate * self.hours_worked



class Freelancer(Employee):

    def __init__(self, employee_id, name, age, project_rate, completed_projects):
        super().__init__(employee_id, name, age)
        self.project_rate = project_rate
        self.completed_projects = completed_projects

    def calculate_salary(self):
        return self.project_rate * self.completed_projects



employees = [

    FullTimeEmployee(1, "Menna", 20, 5000),

    PartTimeEmployee(2, "Mai  ", 21, 100, 80),

    Freelancer(3, "salmaa", 20, 1500, 4),

    FullTimeEmployee(4, "Maryam ", 21, 7000)

]



total_payroll = 0
highest_salary = 0
highest_employee = None

for employee in employees:

    employee.display_info()

    salary = employee.calculate_salary()

    print("Salary:", salary)
    print("--------------------")

    total_payroll += salary

    if salary > highest_salary:
        highest_salary = salary
        highest_employee = employee



print("\n Employee Report --> ")

print("Total number of employees:", len(employees))

print("Total payroll:", total_payroll)

print("Employee with highest salary:", highest_employee.name)

print("Highest salary:", highest_salary)
```
