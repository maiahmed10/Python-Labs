```python
def solution(inputString):
    if inputString == inputString[::-1]:
        return True
    else:
        return False


print(solution("aabaa"))
print(solution("abac"))
print(solution("a"))
```

```python
names = ["Ahmed", "Amani","Sara", "Samar", "Omar", "Ola","mai"]

result = {}

for i in names:
    alpha = i[0]

    if alpha not in result:
        result[alpha] = []

    result[alpha].append(i)

result = dict(sorted(result.items()))

print(result)
```

```python
inputArray = ["aba", "aa", "ad", "vcd", "aba"]

max_length = 0

for word in inputArray:
    if len(word) > max_length:
        max_length = len(word)

result = []

for word in inputArray:
    if len(word) == max_length:
        result.append(word)

print(result)
```

```python
apple = [10, 20, 30]
capacity = [15, 25, 40, 10]

total_apples = sum(apple)

capacity.sort(reverse=True)

total_capacity = 0
boxes = 0

for c in capacity:
    total_capacity += c
    boxes += 1

    if total_capacity >= total_apples:
        break

print(boxes)
```

```python
tasks = []



def add_task():
    task = input("Enter a new task: ")
    tasks.append({"task": task, "done": False})
    print("Task added successfully!")


def view_tasks():
    if len(tasks) == 0:
        print("No tasks found.")
        return

    for i in range(len(tasks)):
        status = "DONE" if tasks[i]["done"] else "NOT DONE"
        print(i + 1, "-", tasks[i]["task"], "-->", status)



def update_task():
    view_tasks()

    index = int(input("Enter task number to update: "))

    if index >= 1 and index <= len(tasks):
        new_task = input("Enter new task: ")
        tasks[index - 1]["task"] = new_task
        print("Task updated successfully!")
    else:
        print("Invalid task number.")

def delete_task():
    view_tasks()

    index = int(input("Enter task number to delete: "))

    if index >= 1 and index <= len(tasks):
        tasks.pop(index - 1)
        print("Task deleted successfully!")
    else:
        print("Invalid task number.")



def mark_done():
    view_tasks()

    index = int(input("Enter task number to mark as done: "))

    if index >= 1 and index <= len(tasks):
        tasks[index - 1]["done"] = True
        print("Task marked as DONE!")
    else:
        print("Invalid task number.")



while True:

    print("\n===== TO-DO APPLICATION =====")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Update Task")
    print("4. Delete Task")
    print("5. Mark Task as Done")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_task()

    elif choice == "2":
        view_tasks()

    elif choice == "3":
        update_task()

    elif choice == "4":
        delete_task()

    elif choice == "5":
        mark_done()

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid choice.")
```
